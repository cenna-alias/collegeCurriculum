**1. Series (1-Dimensional)**

```python
import pandas as pd
s=pd.Series([10,20,30,40])
print(s)
```

**2. DataFrame (2-Dimensional)**

```python
import pandas as pd
data = {
    "Name": ["Alice","Bob"],
    "Age": [20,22]
}
df=pd.DataFrame(data)
print(df)
```

```python
import pandas as pd
data = {
    "Milk price list": [25,50,250],
    "Curd price list": [15,30,350]
}
df=pd.DataFrame(data)
print(df)
```

```python
# Transpose

import pandas as pd

df = pd.DataFrame({
    "A": [1, 2, 3],
    "B": [4, 5, 6]
})
print(df)

df_t = df.T
print(df_t)
```

```python
# Product

import pandas as pd

df = pd.DataFrame({
    "A": [2, 3, 4],
    "B": [5, 6, 7]
})

print(df.prod())
```
